```python
import pandas as pd
from sklearn.feature_selection import VarianceThreshold
X=pd.DataFrame({
    "Constant_feature":[1,1,1,1,1],
    "almost_constant ":[0,0,0,0,1],
    "usefull_feature":[10,20,30,40,50]
})

selector=VarianceThreshold(threshold=0.1)
X_selected=selector.fit_transform(X)
selected_columns =X.columns[selector.get_support()]
print("Selected Features :", list(selected_columns))
print(selected_columns)
```

```python
from sklearn.datasets import load_breast_cancer
from sklearn.feature_selection import SelectKBest, chi2
from sklearn.preprocessing import MinMaxScaler

# 1. Load the data
data = load_breast_cancer()
X = data.data
y = data.target

# 2. Scale features to be non-negative (required for chi2)
X_non_negative = MinMaxScaler().fit_transform(X)

# 3. Select the top 5 features
selector = SelectKBest(score_func=chi2, k=5)
X_selected = selector.fit_transform(X_non_negative, y)

# 4. Get the names of the selected features
selected_features = data.feature_names[selector.get_support()]

print("Selected Features:", selected_features)
```

```python
from sklearn.datasets import load_breast_cancer
from sklearn.feature_selection import SelectKBest,chi2
from sklearn.preprocessing import MinMaxScaler
data=load_breast_cancer()
X=data.data
y=data.target
X_non_negative=MinMaxScaler().fit_transform(X)
selector=SelectKBest(score_func=chi2,k=5)
X_selected=selector.fit_transform(X_non_negative,y)
selected_features=data.feature_names[selector.get_support()]
print("Selected Features :",list(selected_features))
#take the data and check how many columns are there so that we know how do we get 5 best features
```

Perform Variance Threshold For Dataset 1 and Dataset 2


```python
import pandas as pd
from sklearn.feature_selection import VarianceThreshold



df = pd.read_csv("/content/data-1.csv")

print("Original Features:", df.columns.tolist())

# 2. Initialize VarianceThreshold
# threshold=0.0 removes ONLY strictly constant features.
# Here we use 0.5 to remove both Feature A (var=0) and Feature B (var=0.24)
selector = VarianceThreshold(threshold=0.5)

# 3. Fit the selector to the data and transform it
X_selected = selector.fit_transform(df)

# 4. Retrieve the names of the features that were kept
features_kept = df.columns[selector.get_support()]

# 5. Convert back to a Pandas DataFrame for readability
df_selected = pd.DataFrame(X_selected, columns=features_kept)

print("Selected Features:", features_kept.tolist())
print("-" * 30)
print(df_selected)
```

```python
import pandas as pd
from sklearn.feature_selection import VarianceThreshold

# 1. Create a sample dataset
# Feature 'A' is completely constant (variance = 0)
# Feature 'B' has very low variance
# Feature 'C' has high variance

df = pd.read_csv("/content/data-2.csv")

print("Original Features:", df.columns.tolist())

# 2. Initialize VarianceThreshold
# threshold=0.0 removes ONLY strictly constant features.
# Here we use 0.5 to remove both Feature A (var=0) and Feature B (var=0.24)
selector = VarianceThreshold(threshold=0.5)

# 3. Fit the selector to the data and transform it
X_selected = selector.fit_transform(df)

# 4. Retrieve the names of the features that were kept
features_kept = df.columns[selector.get_support()]

# 5. Convert back to a Pandas DataFrame for readability
df_selected = pd.DataFrame(X_selected, columns=features_kept)

print("Selected Features:", features_kept.tolist())
print("-" * 30)
print(df_selected)
```
